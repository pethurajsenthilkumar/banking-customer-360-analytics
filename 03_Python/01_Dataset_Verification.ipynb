```python
print("\nRows with unusual last_transaction_month:")
print(
    df[df["last_transaction_month"] > 100000]
    [["id", "last_transaction_month"]]
    .head(20)
)

```

```python
# Check suspicious columns in detail

check_cols = [
    "married",
    "last_transaction_month",
    "balance",
    "monthly_ir",
    "cluster_group"
]

for col in check_cols:
    print(f"\n========== {col} ==========")
    print("Data Type:", df[col].dtype)
    print("Unique Values:", df[col].nunique())
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())
    print("\nTop 10 Values:")
    print(df[col].value_counts().head(10))


```

```python
# Inspect suspicious records

cols_to_check = [
    "id",
    "married",
    "balance",
    "monthly_ir",
    "last_transaction_month",
    "cluster_group",
    "customer_segment",
    "loyalty_level",
    "digital_behavior",
    "risk_segment"
]

print("Rows where last_transaction_month is unusually high:")
display(
    df.loc[
        df["last_transaction_month"] > 100000000,
        cols_to_check
    ].head(20)
)

```

```python
print("Rows where balance is exactly 1 billion:")
display(
    df.loc[
        df["balance"] == 1000000000,
        cols_to_check
    ].head(20)
)
```

```python
# Check numerical columns for zero and negative values

numeric_cols = df.select_dtypes(include=["number"]).columns

for col in numeric_cols:
    print(f"\n========== {col} ==========")
    print("Zero values:", (df[col] == 0).sum())
    print("Negative values:", (df[col] < 0).sum())
```

```python
# Check date columns

date_cols = ["created_date", "last_active_date"]

for col in date_cols:
    print(f"\n========== {col} ==========")
    print("Data Type:", df[col].dtype)
    print("Missing Values:", df[col].isna().sum())
    print("Unique Values:", df[col].nunique())
    print("Sample Values:")
    print(df[col].head(10).to_list())
```

```python
# Check date validity

for col in ["created_date", "last_active_date"]:
    parsed_dates = pd.to_datetime(
        df[col],
        format="%d/%m/%Y",
        errors="coerce"
    )

    print(f"\n========== {col} ==========")
    print("Invalid dates:", parsed_dates.isna().sum())
    print("Minimum date:", parsed_dates.min())
    print("Maximum date:", parsed_dates.max())
```

```python
# Convert date columns to datetime

df["created_date"] = pd.to_datetime(
    df["created_date"],
    format="%d/%m/%Y"
)

df["last_active_date"] = pd.to_datetime(
    df["last_active_date"],
    format="%d/%m/%Y"
)

print("Date conversion completed.\n")

print(df[["created_date", "last_active_date"]].dtypes)

display(df[["created_date", "last_active_date"]].head())
```

```python
# Check whether last_active_date is always on or after created_date

invalid_date_order = df[
    df["last_active_date"] < df["created_date"]
]

print("Rows where last_active_date is before created_date:",
      len(invalid_date_order))

if len(invalid_date_order) > 0:
    display(
        invalid_date_order[
            ["id", "created_date", "last_active_date"]
        ].head(20)
    )
else:
    print("Date relationship is valid for all customers.")
```

```python
# Calculate days between customer creation and last activity

active_duration_days = (
    df["last_active_date"] - df["created_date"]
).dt.days

print("Minimum active duration:", active_duration_days.min(), "days")
print("Maximum active duration:", active_duration_days.max(), "days")
print("Average active duration:", round(active_duration_days.mean(), 2), "days")
print("Customers with 0 active days:", (active_duration_days == 0).sum())
print("Negative durations:", (active_duration_days < 0).sum())
```

```python
# Check categorical columns for blank or whitespace-only values

categorical_cols = df.select_dtypes(include=["str"]).columns

for col in categorical_cols:
    blank_count = df[col].str.strip().eq("").sum()
    whitespace_count = df[col].str.contains(r"^\s+$", regex=True, na=False).sum()

    print(f"{col}:")
    print("  Blank values:", blank_count)
    print("  Whitespace-only values:", whitespace_count)
```

```python
# Check ID uniqueness

print("Total rows:", len(df))
print("Unique IDs:", df["id"].nunique())
print("Duplicate IDs:", df["id"].duplicated().sum())

if df["id"].is_unique:
    print("ID column is unique.")
else:
    print("ID column contains duplicates.")
```

```python
# Check unique categorical values

categorical_cols = df.select_dtypes(include=["str"]).columns

for col in categorical_cols:
    print(f"\n========== {col} ==========")
    print(df[col].unique())
```

```python
# Check how married values relate to customer characteristics

married_check = (
    df.groupby("married")
      .agg(
          customers=("id", "count"),
          avg_age=("age", "mean"),
          avg_balance=("balance", "mean"),
          avg_credit_score=("credit_sco", "mean"),
          avg_engagement=("engagement_score", "mean")
      )
      .round(2)
)

display(married_check)
```

```python
# Check last_transaction_month value ranges

bins = [
    -1,
    0,
    100,
    1000,
    10000,
    100000,
    1000000,
    10000000,
    100000000,
    600000000
]

labels = [
    "0",
    "1-100",
    "101-1K",
    "1K-10K",
    "10K-100K",
    "100K-1M",
    "1M-10M",
    "10M-100M",
    "100M-600M"
]

transaction_ranges = pd.cut(
    df["last_transaction_month"],
    bins=bins,
    labels=labels
)

print("Distribution of last_transaction_month:")
print(transaction_ranges.value_counts().sort_index())
```

```python
# Compare last_transaction_month with last_active_date

transaction_check = df[
    ["id", "last_transaction_month", "last_active_date"]
].copy()

display(
    transaction_check[
        transaction_check["last_transaction_month"] > 0
    ].head(20)
)
```

```python
# Check last transaction values for each last-active year

transaction_check["last_active_year"] = (
    transaction_check["last_active_date"].dt.year
)

display(
    transaction_check[
        transaction_check["last_transaction_month"] > 0
    ]
    .groupby("last_active_year")["last_transaction_month"]
    .agg(["count", "min", "max", "mean"])
    .round(2)
)
```

```python
# Check relationship between last_transaction_month and digital_behavior

print(
    pd.crosstab(
        df["digital_behavior"],
        df["last_transaction_month"] > 0
    )
)
```

```python
# Distribution of last_transaction_month for mobile customers

mobile_transactions = df[
    df["digital_behavior"] == "mobile"
]["last_transaction_month"]

print("Mobile customers:", len(mobile_transactions))
print("Unique values:", mobile_transactions.nunique())
print("Minimum:", mobile_transactions.min())
print("Maximum:", mobile_transactions.max())
print("Mean:", round(mobile_transactions.mean(), 2))
print("\nTop 15 values:")
print(mobile_transactions.value_counts().head(15))
```

```python
# Check relationships between last_transaction_month
# and other important numeric fields

cols = [
    "last_transaction_month",
    "balance",
    "monthly_ir",
    "engagement_score",
    "risk_score"
]

print(df[cols].corr()["last_transaction_month"].sort_values(ascending=False))
```

```python
# Compare engagement score across last_transaction_month ranges

df["transaction_range"] = pd.cut(
    df["last_transaction_month"],
    bins=[
        -1, 0, 1_000_000, 10_000_000,
        100_000_000, 600_000_000
    ],
    labels=[
        "0",
        "1M or less",
        "1M-10M",
        "10M-100M",
        "100M-600M"
    ]
)

print(
    df.groupby(
        "transaction_range",
        observed=True
    )["engagement_score"]
    .agg(["count", "mean", "min", "max"])
    .round(2)
)
```

```python
# Remove temporary analysis column

df.drop(columns=["transaction_range"], inplace=True)

print("Temporary column removed.")
print("Current shape:", df.shape)
```

```python
print("=== FINAL NUMERIC RANGE CHECK ===")

checks = {
    "credit_sco": (300, 900),
    "age": (0, 120),
    "tenure_ye": (0, 100),
    "nums_card": (0, 20),
    "nums_service": (0, 20),
    "engagement_score": (0, 100),
    "risk_score": (0, 1),
    "cluster_group": (1, 10)
}

for col, (min_val, max_val) in checks.items():
    below = (df[col] < min_val).sum()
    above = (df[col] > max_val).sum()

    print(
        f"{col}: below_min={below}, "
        f"above_max={above}"
    )
```
